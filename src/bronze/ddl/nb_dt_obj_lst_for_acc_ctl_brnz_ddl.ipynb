#### DDL for Bronze Table "access_control_data_object_list_ing"

```python
access_control_data_object_list_ing_query = f"""CREATE TABLE IF NOT EXISTS {catalog}.ing_demo.access_control_data_object_list_ing (
    table_catalog STRING NOT NULL COMMENT 'Catalog of the data object on which access control is applied',
    table_schema STRING NOT NULL COMMENT 'Schema of the data object on which access control is applied',
    table_name STRING NOT NULL COMMENT 'Name of the data object on which access control is applied',
    business_entity STRING COMMENT 'Name of the business entity under which the table belongs to. Like - Finance, GPAS etc.',
    table_type STRING NOT NULL COMMENT 'Type of the data object on which access control is applied. Possible values : TABLE / MV',
    control_column_name STRING NOT NULL COMMENT 'Name of the secured column of the data object on which access control is applied',
    active_flag STRING NOT NULL COMMENT 'Denotes if the record is active',
    created_by STRING COMMENT 'The Databricks user who created the record',
    created_date TIMESTAMP COMMENT 'The date and time when the record was created',
    last_updated_date TIMESTAMP COMMENT 'The date and time when the record was last updated',
    PRIMARY KEY (dl_name) 
)
TBLPROPERTIES (
    'comment' = 'This table contains the list of data objects with information about associated row level filters',
    'tag.domain_name' = 'demo',
    'tag.sub_domain_code' = 'dw',
    'tag.table_type' = 'control table',
    'tag.data_objects' = 'List of Data Objects with Row Level Filters',
    'delta.autoOptimize.optimizeWrite' = 'true',
    'delta.enableChangeDataFeed' = 'true',
    'delta.enableDeletionVectors' = 'true',
    'delta.feature.allowColumnDefaults' = 'supported',
    'delta.feature.changeDataFeed' = 'supported',
    'delta.feature.deletionVectors' = 'supported',
    'delta.minReaderVersion' = '3',
    'delta.minWriterVersion' = '7')"""

spark.sql(access_control_data_object_list_ing_query)
```
