```python
import json
```

#### Fetch Parameter from Job

```python
environment = dbutils.widgets.get("environment")
```

#### Read the "config.json" File and Return the Object as Python Dictionary Object

```python
def read_config_json ():
    try:
        # Open the File in Read Mode ('r')
        with open('config.json', 'r') as json_file:
            # Load the JSON Data into a Python Dictionary
            json_data = json.load(json_file)
            # Loop Through Each Objects from the "configs", i.e., Array of JSON Objects
            for config in json_data["configs"]:
                if config["environment"].lower() == environment.lower():
                    dict_config = config["parameters"]
        return dict_config
    except FileNotFoundError as file_not_found_ex:
        print("Error: The File 'config.json' Was Not Found.")
        raise(file_not_found_ex)
    except json.JSONDecodeError as json_decoder_ex:
        print("Error: Failed to Decode JSON from the File (Invalid JSON Format).")
        raise(json_decoder_ex)
```

#### Create Variables from the Python Dictionary Object

```python
# Call the Function "read_config_json" and Assign the Return Value into a Python Dictionary
dict_config = read_config_json ()

# Read the Python Dictionary and Assign Each JSON Object into Each Variable
catalog = dict_config.get('catalog')
```
